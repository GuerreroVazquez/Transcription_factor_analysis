# Add the tissues to selected genes

```python
import pandas as pd
```

```python
selected_genes = pd.read_csv("Selected_genes_orthologs_pathway_lfc.csv", index_col=0)
tissues_df = pd.read_csv("mitoCarta_terms_tissues_only.csv", index_col=0)
```

```python
tissues_df[tissues_df['Tissues']!='all 14'].head(3)
```

```python
selected_genes_tissue = selected_genes.merge(tissues_df, left_index=True, right_index=True, how="left")
```

```python
selected_genes_tissue.to_csv('Selected_genes_orthologs_pathway_lfc_tissues.csv')
```
